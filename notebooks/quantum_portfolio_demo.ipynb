# Quantum-Enhanced Multi-Asset Portfolio Optimization
### Vanguard Quantum Internship Project

This notebook demonstrates a hybrid quantum-classical pipeline for multi-asset portfolio optimization. We compare a classical Mean-Variance baseline (CVXPY) against a Quantum Quadratic Unconstrained Binary Optimization (QUBO) algorithm implemented with Qiskit.

```python
import sys
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Add project root to Python path
sys.path.append(os.path.abspath('..'))

from data.generate_data import save_synthetic_data
from optimizer.classical_optimizer import ClassicalPortfolioOptimizer
from optimizer.quantum_optimizer import QuantumPortfolioOptimizer
from optimizer.portfolio_optimizer import UnifiedPortfolioEngine
from copilot.explanation_engine import PortfolioCoPilot

print('All modules loaded successfully!')
```

## 1. Asset Universe & Covariance Matrix
We evaluate an 8-asset portfolio containing Growth Equities, Fixed Income, and Commodities.

```python
asset_df, cov_df = save_synthetic_data('../data')
display(asset_df)
```

## 2. Classical Optimization (CVXPY Baseline)
Continuous Mean-Variance Markowitz optimization using ECOS/SCS interior-point solvers.

```python
classical_opt = ClassicalPortfolioOptimizer(asset_df, cov_df)
classical_res = classical_opt.optimize(risk_aversion=0.5, max_weight=0.40)
print('Classical Expected Return:', f"{classical_res['expected_return']*100:.2f}%")
print('Classical Portfolio Risk:', f"{classical_res['portfolio_risk']*100:.2f}%")
print('Classical Sharpe Ratio:', classical_res['sharpe_ratio'])
```

## 3. Quantum QUBO & Qiskit QAOA Optimization
We discretize decisions into binary variables and minimize $f(x) = x^T Q x$.

```python
quantum_opt = QuantumPortfolioOptimizer(asset_df, cov_df)
quantum_res = quantum_opt.optimize(risk_aversion=0.5, max_weight=0.40)
print('Quantum Expected Return:', f"{quantum_res['expected_return']*100:.2f}%")
print('Quantum Portfolio Risk:', f"{quantum_res['portfolio_risk']*100:.2f}%")
print('Quantum Sharpe Ratio:', quantum_res['sharpe_ratio'])
```

## 4. Empirical Benchmarking
Comparing Classical vs Quantum Solvers.

```python
engine = UnifiedPortfolioEngine('../data')
results = engine.run_optimization(investment_amount=1000000)
benchmarks_df = pd.DataFrame(results['benchmarks'])
display(benchmarks_df)
```

## 5. Portfolio Co-Pilot Explanations
Interpretable explanations for portfolio allocation decisions.

```python
copilot = PortfolioCoPilot(asset_df)
explanations = copilot.generate_portfolio_explanation(results)
print('Summary:', explanations['summary_text'])
print('\nKey Insights:')
for insight in explanations['key_insights']:
    print('-', insight)
```
